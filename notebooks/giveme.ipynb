```python

import subprocess
import re
from pathlib import Path

# Keep recovered files outside the Git repository
recovery = Path.home() / "Desktop" / "RETAILMIND_RECOVERY"
recovery.mkdir(parents=True, exist_ok=True)

result = subprocess.run(
    ["git", "fsck", "--full", "--no-reflogs", "--unreachable"],
    capture_output=True,
    text=True
)

hashes = re.findall(
    r"unreachable blob ([0-9a-f]{40})",
    result.stdout
)

print("Unreachable blobs found:", len(hashes))

for i, blob_hash in enumerate(hashes, 1):
    data = subprocess.run(
        ["git", "cat-file", "blob", blob_hash],
        capture_output=True,
        check=True
    ).stdout

    output = recovery / blob_hash
    output.write_bytes(data)

    if i % 50 == 0:
        print("Recovered:", i)

print("Recovery completed!")
print("Files saved at:", recovery)
```

```python

from pathlib import Path
import json
import shutil

RECOVERY_DIR = Path.home() / "Desktop" / "RETAILMIND_RECOVERY"

OUTPUT_DIR = RECOVERY_DIR / "RECOVERED_NOTEBOOKS"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

day10_candidates = []
day11_candidates = []

for file in RECOVERY_DIR.iterdir():

    if not file.is_file():
        continue

    # Avoid scanning our output folder
    try:
        content = file.read_text(encoding="utf-8")
        notebook = json.loads(content)

        if not isinstance(notebook, dict) or "cells" not in notebook:
            continue

        text = ""

        for cell in notebook["cells"]:
            source = cell.get("source", [])

            if isinstance(source, list):
                text += "\n" + "".join(source)
            else:
                text += "\n" + str(source)

        text = text.lower()

        # Day 10: Seq2Seq + Attention
        terms10 = [
            "seq2seq",
            "sequence to sequence",
            "attention",
            "forecasting",
            "encoder",
            "decoder",
            "lstm"
        ]

        score10 = sum(term in text for term in terms10)

        # Day 11: GAN
        terms11 = [
            "generative adversarial",
            "generator",
            "discriminator",
            "adversarial",
            "generated images",
            "dcgan"
        ]

        score11 = sum(term in text for term in terms11)

        if score10 >= 4:
            destination = OUTPUT_DIR / (
                f"Day10_candidate_{file.name}.ipynb"
            )

            shutil.copy2(file, destination)
            day10_candidates.append((score10, destination.name))

        if score11 >= 3:
            destination = OUTPUT_DIR / (
                f"Day11_candidate_{file.name}.ipynb"
            )

            shutil.copy2(file, destination)
            day11_candidates.append((score11, destination.name))

    except (UnicodeDecodeError, json.JSONDecodeError, OSError):
        continue

print("\n========== DAY 10 CANDIDATES ==========")

for score, name in sorted(day10_candidates, reverse=True):
    print(score, name)

print("\n========== DAY 11 CANDIDATES ==========")

for score, name in sorted(day11_candidates, reverse=True):
    print(score, name)

print("\nTotal Day 10:", len(day10_candidates))
print("Total Day 11:", len(day11_candidates))
print("Saved at:", OUTPUT_DIR)
```
